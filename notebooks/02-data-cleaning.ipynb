# 02 — Data Cleaning
**E-Commerce Sales Analysis**

Cleaning the raw dataset — removing invalid records, fixing types, creating new metrics.

```python
import os
import pandas as pd

# ── Paths ──────────────────────────────────────────────
ROOT       = os.path.abspath("..")
RAW_PATH   = os.path.join(ROOT, "data", "raw", "ecommerce.csv")        # смени ако трябва
CLEAN_PATH = os.path.join(ROOT, "data", "processed", "cleaned_data.csv")

print("Clean file will be saved to:", CLEAN_PATH)
```

```python
# ── Load raw data ──────────────────────────────────────
df = pd.read_csv(RAW_PATH, encoding="latin-1")
print(f"Raw shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
# ── 1. Fix date type ───────────────────────────────────
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
print("InvoiceDate dtype:", df['InvoiceDate'].dtype)
```

```python
# ── 2. Remove missing CustomerID ───────────────────────
before = len(df)
df = df.dropna(subset=['CustomerID'])
print(f"Removed {before - len(df):,} rows with missing CustomerID")
print(f"Remaining: {len(df):,} rows")
```

```python
# ── 3. Remove returns (negative quantity) ──────────────
before = len(df)
df = df[df['Quantity'] > 0]
print(f"Removed {before - len(df):,} return rows (negative quantity)")
print(f"Remaining: {len(df):,} rows")
```

```python
# ── 4. Remove invalid prices ───────────────────────────
before = len(df)
df = df[df['UnitPrice'] > 0]
print(f"Removed {before - len(df):,} rows with zero/negative price")
print(f"Remaining: {len(df):,} rows")
```

```python
# ── 5. Remove duplicate rows ───────────────────────────
before = len(df)
df = df.drop_duplicates()
print(f"Removed {before - len(df):,} duplicate rows")
print(f"Remaining: {len(df):,} rows")
```

```python
# ── 6. Create TotalPrice column ────────────────────────
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']
print("TotalPrice column created")
df[['Quantity', 'UnitPrice', 'TotalPrice']].head()
```

```python
# ── 7. Extract date parts ──────────────────────────────
df['Year']  = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['Day']   = df['InvoiceDate'].dt.day_name()
df['Hour']  = df['InvoiceDate'].dt.hour

print("Date columns created: Year, Month, Day, Hour")
df[['InvoiceDate', 'Year', 'Month', 'Day', 'Hour']].head()
```

```python
# ── 8. Fix CustomerID type ─────────────────────────────
df['CustomerID'] = df['CustomerID'].astype(int).astype(str)
print("CustomerID converted to string")
```

```python
# ── Final check ────────────────────────────────────────
print(f"\nFinal shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
print(f"Missing values remaining:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# ── Save cleaned data ──────────────────────────────────
os.makedirs(os.path.dirname(CLEAN_PATH), exist_ok=True)

df_sample = df.sample(n=min(10000, len(df)), random_state=42)
df_sample.to_csv(CLEAN_PATH, index=False)

print(f"Saved {len(df_sample):,} rows to: {CLEAN_PATH}")
```
