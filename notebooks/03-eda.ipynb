# 03 — Exploratory Data Analysis (EDA)
**E-Commerce Sales Analysis**

Visual exploration of the cleaned dataset — revenue trends, top products, country performance, and customer behavior.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# ── Paths ──────────────────────────────────────────────
ROOT        = os.path.abspath("..")
CLEAN_PATH  = os.path.join(ROOT, "data", "processed", "cleaned_data.csv")
FIGURES_DIR = os.path.join(ROOT, "reports", "figures")
os.makedirs(FIGURES_DIR, exist_ok=True)

# ── Style ──────────────────────────────────────────────
plt.rcParams.update({
    "figure.figsize"  : (12, 5),
    "axes.spines.top" : False,
    "axes.spines.right": False,
    "axes.grid"       : True,
    "grid.alpha"      : 0.3,
    "font.size"       : 11,
})

print("Setup complete")
```

```python
# ── Load cleaned data ──────────────────────────────────
df = pd.read_csv(CLEAN_PATH, parse_dates=["InvoiceDate"])
print(f"Loaded: {df.shape[0]:,} rows × {df.shape[1]} columns")
df.head(3)
```

## 1 — Monthly Revenue Trend

```python
monthly = (
    df.groupby(df['InvoiceDate'].dt.to_period('M'))['TotalPrice']
    .sum()
    .reset_index()
)
monthly['InvoiceDate'] = monthly['InvoiceDate'].astype(str)

fig, ax = plt.subplots()
ax.plot(monthly['InvoiceDate'], monthly['TotalPrice'], marker='o', linewidth=2, color='#2563eb')
ax.fill_between(range(len(monthly)), monthly['TotalPrice'], alpha=0.1, color='#2563eb')
ax.set_title('Monthly Revenue', fontsize=14, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('Revenue (£)')
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'£{x:,.0f}'))
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, 'monthly_revenue.png'), dpi=150)
plt.show()
print("Saved: monthly_revenue.png")
```

## 2 — Top 10 Products by Revenue

```python
top_products = (
    df.groupby('Description')['TotalPrice']
    .sum()
    .sort_values(ascending=True)
    .tail(10)
)

fig, ax = plt.subplots()
bars = ax.barh(top_products.index, top_products.values, color='#2563eb')
ax.bar_label(bars, fmt='£{:,.0f}', padding=5, fontsize=9)
ax.set_title('Top 10 Products by Revenue', fontsize=14, fontweight='bold')
ax.set_xlabel('Revenue (£)')
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'£{x:,.0f}'))
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, 'top_products.png'), dpi=150)
plt.show()
print("Saved: top_products.png")
```

## 3 — Revenue by Country (Top 10)

```python
top_countries = (
    df.groupby('Country')['TotalPrice']
    .sum()
    .sort_values(ascending=True)
    .tail(10)
)

fig, ax = plt.subplots()
colors = ['#2563eb' if c == 'United Kingdom' else '#93c5fd' for c in top_countries.index]
bars = ax.barh(top_countries.index, top_countries.values, color=colors)
ax.bar_label(bars, fmt='£{:,.0f}', padding=5, fontsize=9)
ax.set_title('Top 10 Countries by Revenue', fontsize=14, fontweight='bold')
ax.set_xlabel('Revenue (£)')
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'£{x:,.0f}'))
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, 'revenue_by_country.png'), dpi=150)
plt.show()
print("Saved: revenue_by_country.png")
```

## 4 — Orders by Day of Week

```python
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
by_day = df.groupby('Day')['TotalPrice'].sum().reindex(day_order)

fig, ax = plt.subplots()
ax.bar(by_day.index, by_day.values, color='#2563eb')
ax.set_title('Revenue by Day of Week', fontsize=14, fontweight='bold')
ax.set_ylabel('Revenue (£)')
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'£{x:,.0f}'))
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, 'revenue_by_day.png'), dpi=150)
plt.show()
print("Saved: revenue_by_day.png")
```

## 5 — Revenue by Hour

```python
by_hour = df.groupby('Hour')['TotalPrice'].sum()

fig, ax = plt.subplots()
ax.plot(by_hour.index, by_hour.values, marker='o', linewidth=2, color='#2563eb')
ax.set_title('Revenue by Hour of Day', fontsize=14, fontweight='bold')
ax.set_xlabel('Hour')
ax.set_ylabel('Revenue (£)')
ax.set_xticks(range(0, 24))
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'£{x:,.0f}'))
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, 'revenue_by_hour.png'), dpi=150)
plt.show()
print("Saved: revenue_by_hour.png")
```

## 📝 EDA Summary

| Chart | Key Insight |
|---|---|
| Monthly Revenue | Strong Q4 seasonality — peak in November |
| Top Products | Top 10 products drive significant share of revenue |
| By Country | UK dominates; Netherlands and Germany follow |
| By Day | Thursday and Tuesday are the strongest days |
| By Hour | Peak activity between 10:00 and 15:00 |
