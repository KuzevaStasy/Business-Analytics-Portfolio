# 01 — Data Overview
**E-Commerce Sales Analysis**

First look at the raw dataset — structure, size, data types, and missing values.

```python
import os
import pandas as pd

# ── Paths ──────────────────────────────────────────────
ROOT      = os.path.abspath("..")
RAW_PATH  = os.path.join(ROOT, "data", "raw", "ecommerce.csv")

print("Root:", ROOT)
print("Data file:", RAW_PATH)
```

```python
# ── Load data ──────────────────────────────────────────
df = pd.read_csv(RAW_PATH, encoding="latin-1")

print(f"Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
# ── First look ─────────────────────────────────────────
df.head(10)
```

```python
# ── Column types ───────────────────────────────────────
df.dtypes
```

```python
# ── Missing values ─────────────────────────────────────
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

pd.DataFrame({
    "Missing count" : missing,
    "Missing %"     : missing_pct
}).query("`Missing count` > 0")
```

```python
# ── Basic stats ────────────────────────────────────────
df.describe()
```

```python
# ── Unique values per column ───────────────────────────
pd.DataFrame({
    "Unique values": df.nunique()
})
```

```python
# ── Countries in dataset ───────────────────────────────
print(f"Total countries: {df['Country'].nunique()}")
df['Country'].value_counts().head(10)
```

```python
# ── Date range ─────────────────────────────────────────
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

print(f"From : {df['InvoiceDate'].min()}")
print(f"To   : {df['InvoiceDate'].max()}")
```
